```python
class XLWICExample:
    def __init__(self, word, sentence1, sentence2, postag, pos1, pos2, label):
        self.word = word
        self.sentence1 = sentence1
        self.sentence2 = sentence2
        self.postag = postag
        self.pos1 = pos1
        self.pos2 = pos2
        self.label = label

    def __repr__(self):
        return (
            f"WordAlignmentExample(word={self.word}, "
            f"pos1={self.pos1}, pos2={self.pos2})"
        )
```

```python
# gold_labels = []
with open(xlwic_gold_path, encoding="utf8") as gold_file:
    gold_labels = list(map(lambda line: int(line.strip('\n')), gold_file.readlines()))
```

```python
dataset = []
with open(xlwic_path, encoding="utf8") as truth_file:
	for i, line in enumerate(truth_file.readlines()):
		example = line.strip().split("\t")
		dataset.append(XLWICExample(word=example[0], 
							  sentence1=example[6], 
							  sentence2=example[7],
							  postag=example[1], 
							  pos1=(example[2], example[3]),
							  pos2=(example[4], example[5]),
							  label=gold_labels[i]))
```
